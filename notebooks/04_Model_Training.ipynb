# 04 - Model Training with YOLOv8n (50 Epochs Baseline)
## Proyek: RoadDamage_AI

### Konsep Pelatihan 50 Epoch
Pada tahap ini, kita melatih model **YOLOv8n (Nano)** dari Ultralytics selama **50 Epoch** (dengan early stopping patience=10):
- **Pretrained Backbone:** `yolov8n.pt` (Pretrained pada MS COCO) untuk Transfer Learning.
- **Ukuran Input Citra:** 640 x 640 pixel.
- **Dataset:** `data/data.yaml` (1,406 citra Train, 401 citra Val).
- **Target Bobot Model:** `backend/models/road_damage.pt`.


```python
import os
import shutil
from ultralytics import YOLO

DATA_YAML = os.path.abspath("../data/data.yaml")
BACKEND_MODEL_DIR = os.path.abspath("../backend/models")
os.makedirs(BACKEND_MODEL_DIR, exist_ok=True)

print(f"Dataset YAML: {DATA_YAML}")
print(f"Backend Model Dir: {BACKEND_MODEL_DIR}")
```

---
### 1. Inisialisasi Pretrained Model YOLOv8n


```python
model = YOLO("yolov8n.pt")
print(f"Loaded base model: {model.model_name}")
```

---
### 2. Status dan Konfigurasi Training 50 Epoch


```python
RUNS_TRAIN_DIR = os.path.abspath("../runs/detect/train_50epochs")
best_weight_path = os.path.join(RUNS_TRAIN_DIR, "weights", "best.pt")
csv_path = os.path.join(RUNS_TRAIN_DIR, "results.csv")

if os.path.exists(best_weight_path):
    print(f"Hasil training 50 epoch tersedia di: {RUNS_TRAIN_DIR}")
    print(f"Best weight terverifikasi: {best_weight_path}")
    print(f"Ukuran file: {os.path.getsize(best_weight_path)/(1024*1024):.2f} MB")
else:
    print("Memulai training 50 epochs...")
    results = model.train(
        data=DATA_YAML,
        epochs=50,
        patience=10,
        imgsz=640,
        batch=16,
        seed=42,
        project=os.path.abspath("../runs/detect"),
        name="train_50epochs",
        exist_ok=True,
        verbose=True
    )
```

---
### 3. Menyimpan Model Terbaik ke Backend Service


```python
target_model_path = os.path.join(BACKEND_MODEL_DIR, "road_damage.pt")

if os.path.exists(best_weight_path):
    shutil.copyfile(best_weight_path, target_model_path)
    print(f"Model terbaik 50-epoch berhasil disimpan di: {target_model_path}")
    print(f"Ukuran file model: {os.path.getsize(target_model_path) / (1024*1024):.2f} MB")
else:
    print("File best.pt belum ditemukan.")
```

---
### 4. Kesimpulan Training 50 Epoch
Pelatihan baseline 50 epoch selesai dengan konvergensi yang jauh lebih matang dibandingkan 4 epoch awal. Bobot `road_damage.pt` siap dievaluasi pada `05_Model_Evaluation.ipynb`.
