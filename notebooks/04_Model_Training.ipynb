# 04 - Model Training with YOLOv8n
## Proyek: RoadDamage_AI

### Konsep Pelatihan
Pada tahap ini, kita menggunakan model **YOLOv8n (Nano)** dari Ultralytics sebagai arsitektur dasar.
- **Pretrained Weights:** `yolov8n.pt` (Pretrained pada MS COCO) untuk teknik Transfer Learning.
- **Ukuran Model:** Sangat ringan (~3.2 Juta parameter, ~6.5 MB), cocok untuk efisiensi komputasi dan mencegah overfitting parah pada dataset berukuran 2000 gambar.
- **Ukuran Input Gambar:** 640 x 640 pixel.
- **Target Penyimpanan:** `backend/models/road_damage.pt`.


```python
import os
import shutil
from ultralytics import YOLO

# Path konfigurasi dataset
DATA_YAML = os.path.abspath("../data/data.yaml")
BACKEND_MODEL_DIR = os.path.abspath("../backend/models")
os.makedirs(BACKEND_MODEL_DIR, exist_ok=True)

print(f"Dataset YAML: {DATA_YAML}")
print(f"Backend Model Dir: {BACKEND_MODEL_DIR}")
```

---
### 1. Inisialisasi Pretrained Model YOLOv8n


```python
# Memuat model pretrained YOLOv8n
model = YOLO("yolov8n.pt")
print(f"Model Name: {model.model_name}")
```

---
### 2. Memulai / Memverifikasi Proses Training


```python
# Path checkpoint hasil training
RUNS_TRAIN_DIR = os.path.abspath("../runs/detect/train")
best_weight_path = os.path.join(RUNS_TRAIN_DIR, "weights", "best.pt")

if not os.path.exists(best_weight_path):
    print("Memulai training YOLOv8n...")
    results = model.train(
        data=DATA_YAML,
        epochs=3,
        imgsz=640,
        batch=16,
        seed=42,
        project=os.path.abspath("../runs/detect"),
        name="train",
        exist_ok=True,
        verbose=True
    )
else:
    print(f"Hasil training YOLOv8n tersedia di: {RUNS_TRAIN_DIR}")
    print(f"Best weight terdeteksi: {best_weight_path}")
```

---
### 3. Menyimpan Model Terbaik ke Backend Service


```python
target_model_path = os.path.join(BACKEND_MODEL_DIR, "road_damage.pt")

if os.path.exists(best_weight_path):
    shutil.copyfile(best_weight_path, target_model_path)
    print(f"Model terbaik berhasil disimpan di: {target_model_path}")
    print(f"Ukuran file model: {os.path.getsize(target_model_path) / (1024*1024):.2f} MB")
else:
    print("File best.pt belum ditemukan.")
```
